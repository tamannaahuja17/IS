```python
def encryption_function(text, key):
    text = text.replace(" ", "")
    cols = len(key)
    rows = (len(text) + cols - 1) // cols
    matrix = [['X'] * cols for _ in range(rows)]
    k = 0
    for i in range(rows):
        for j in range(cols):
            if k < len(text):
                matrix[i][j] = text[k]
                k += 1
    result = ""
    for num in range(1, cols+1):
        col_index = key.index(num)
        for row in matrix:
            result += row[col_index]
    return result
```

```python
def decryption_function(cipher, key):
    cols = len(key)
    rows = len(cipher) // cols
    matrix = [[''] * cols for _ in range(rows)]
    curr = 0
    for num in range(1, cols + 1):
        col_index = key.index(num)
        for r in range(rows):
            matrix[r][col_index] = cipher[curr]
            curr += 1
    plain = ""
    for r in range(rows):
        for c in range(cols):
            plain += matrix[r][c]
    return plain
```

```python
text = input("text: ")
key = list(map(int, input("key: ").split()))

enc = encryption_function(text, key)
print("Encoded:", enc)

dec = decryption_function(enc, key)
print("Decoded:", dec)
```
