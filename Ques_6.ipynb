## Monoalphabetic Cipher

```python
def monoalphabetic_encrypt(plaintext, key):
    encrypted_text = ""

    alphabet = 'abcdefghijklmnopqrstuvwxyz'
    mapping = str.maketrans(alphabet, key)

    for char in plaintext:
        if char.isalpha():
            if char.islower():
                encrypted_text += char.translate(mapping)
            else:
                encrypted_text += char.lower().translate(mapping).upper()
        else:
            encrypted_text += char
    return encrypted_text

def monoalphabetic_decrypt(ciphertext, key):
    decrypted_text = ""

    alphabet = 'abcdefghijklmnopqrstuvwxyz'
    reverse_mapping = str.maketrans(key, alphabet)

    for char in ciphertext:
        if char.isalpha():
            if char.islower():
                decrypted_text += char.translate(reverse_mapping)
            else:
                decrypted_text += char.lower().translate(reverse_mapping).upper()
        else:
            decrypted_text += char
    return decrypted_text

# Example Usage - Monoalphabetic Cipher
mono_plaintext = "Hello World"
mono_key = "zyxwuvtsrqponmlkjihgfedcba" # A simple reverse alphabet key

mono_encrypted = monoalphabetic_encrypt(mono_plaintext, mono_key)
mono_decrypted = monoalphabetic_decrypt(mono_encrypted, mono_key)

print(f"Monoalphabetic Plaintext: {mono_plaintext}")
print(f"Monoalphabetic Key: {mono_key}")
print(f"Monoalphabetic Encrypted: {mono_encrypted}")
print(f"Monoalphabetic Decrypted: {mono_decrypted}")
```

## Polyalphabetic Cipher (Vigenere Cipher)

```python
def polyalphabetic_encrypt(plaintext, key):
    encrypted_text = []
    key_len = len(key)
    key_as_int = [ord(char) for char in key.lower()]

    for i, char in enumerate(plaintext):
        if char.isalpha():
            shift = key_as_int[i % key_len] - ord('a')
            start = ord('a') if char.islower() else ord('A')
            encrypted_char_code = start + (ord(char) - start + shift) % 26
            encrypted_text.append(chr(encrypted_char_code))
        else:
            encrypted_text.append(char)
    return "".join(encrypted_text)

def polyalphabetic_decrypt(ciphertext, key):
    decrypted_text = []
    key_len = len(key)
    key_as_int = [ord(char) for char in key.lower()]

    for i, char in enumerate(ciphertext):
        if char.isalpha():
            shift = key_as_int[i % key_len] - ord('a')
            start = ord('a') if char.islower() else ord('A')
            decrypted_char_code = start + (ord(char) - start - shift + 26) % 26 # +26 to handle negative results of modulo
            decrypted_text.append(chr(decrypted_char_code))
        else:
            decrypted_text.append(char)
    return "".join(decrypted_text)


poly_plaintext = "Attack at dawn"
poly_key = "Lemon"

poly_encrypted = polyalphabetic_encrypt(poly_plaintext, poly_key)
poly_decrypted = polyalphabetic_decrypt(poly_encrypted, poly_key)

print(f"\nPolyalphabetic Plaintext: {poly_plaintext}")
print(f"Polyalphabetic Key: {poly_key}")
print(f"Polyalphabetic Encrypted: {poly_encrypted}")
print(f"Polyalphabetic Decrypted: {poly_decrypted}")
```
