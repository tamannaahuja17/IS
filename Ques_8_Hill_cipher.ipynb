```python
import numpy as np

def modInverse(a, m):
    for x in range(1, m):
        if (((a % m) * (x % m)) % m == 1):
            return x
    return -1

def hill_encrypt(text, key):
    text = text.lower().replace(" ", "")
    if len(text) % 2 != 0:
        text += 'x'
    result = ""
    for i in range(0, len(text), 2):
        x = ord(text[i]) - ord('a')
        y = ord(text[i+1]) - ord('a')
        c1 = (key[0][0]*x + key[0][1]*y) % 26
        c2 = (key[1][0]*x + key[1][1]*y) % 26
        result += chr(c1 + ord('a'))
        result += chr(c2 + ord('a'))
    return result

def hill_decrypt(cipher, key):

    det = (key[0][0]*key[1][1] - key[0][1]*key[1][0]) % 26
    inv_det = modInverse(det, 26)
    if inv_det == -1:
        return "Error: Matrix is not invertible modulo 26"


    inv_key = [
        [(key[1][1] * inv_det) % 26, (-key[0][1] * inv_det) % 26],
        [(-key[1][0] * inv_det) % 26, (key[0][0] * inv_det) % 26]
    ]

    result = ""
    for i in range(0, len(cipher), 2):
        x = ord(cipher[i]) - ord('a')
        y = ord(cipher[i+1]) - ord('a')
        p1 = (inv_key[0][0]*x + inv_key[0][1]*y) % 26
        p2 = (inv_key[1][0]*x + inv_key[1][1]*y) % 26
        result += chr(int(p1) + ord('a'))
        result += chr(int(p2) + ord('a'))
    return result


choice = input("Use hardcoded key (3, 3, 2, 5)? (y/n): ").lower()
if choice == 'y':
    key = [[3, 3], [2, 5]]
else:
    print("Enter 4 values for 2x2 matrix (e.g., 3 3 2 5):")
    a, b, c, d = map(int, input().split())
    key = [[a, b], [c, d]]

text = input("Enter text: ")
encrypted = hill_encrypt(text, key)
print("Encrypted:", encrypted)
decrypted = hill_decrypt(encrypted, key)
print("Decrypted:", decrypted)
```
